```python
import pandas as pd
import nltk
import string

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from textblob import TextBlob
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('punkt_tab')
```

```python
data = {
    "Review": [
        "The room was clean and spacious. Staff were very friendly.",
        "Food was delicious and service was excellent.",
        "The room was dirty and bathroom smelled bad.",
        "Staff were rude and check-in took too long.",
        "Hotel location is good but WiFi is slow.",
        "Breakfast was average and rooms were comfortable.",
        "AC was not working properly.",
        "Swimming pool was clean and well maintained."
    ]
}

df = pd.DataFrame(data)
display(df)
```

```python
lemmatizer = WordNetLemmatizer()

def preprocess(text):

    # Convert to lowercase
    text = text.lower()

    # Remove punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))

    # Tokenize
    words = word_tokenize(text)

    # Remove stop words
    stop_words = set(stopwords.words('english'))
    words = [word for word in words if word not in stop_words]

    # Lemmatization
    words = [lemmatizer.lemmatize(word) for word in words]

    return " ".join(words)

# Let's test the function with an example sentence to ensure it works
sample_text = "NLTK is a powerful library for natural language processing, and it's quite useful!"
cleaned_sample_text = preprocess(sample_text)
print(f"Original text: {sample_text}")
print(f"Cleaned text: {cleaned_sample_text}")
```

```python
df["Processed Review"] = df["Review"].apply(preprocess)
display(df)
```

```python
def get_sentiment(text):
    score = TextBlob(text).sentiment.polarity
    if score > 0:
        return "Positive"
    elif score < 0:
        return "Negative"
    else:
        return "Neutral"

df["Sentiment"] = df["Processed Review"].apply(get_sentiment)
display(df)
```

```python
issues = {
    "Room": ["room", "bed", "bathroom", "ac"],
    "Staff": ["staff", "service", "reception"],
    "Food": ["food", "breakfast", "restaurant"],
    "Cleanliness": ["clean", "dirty", "smell"],
    "Amenities": ["wifi", "pool", "gym"]
}

def get_issue(processed_review):
    for issue_type, keywords in issues.items():
        for keyword in keywords:
            if keyword in processed_review:
                return issue_type
    return "Other"

df["Issue"] = df["Processed Review"].apply(get_issue)
display(df)
```

```python
sentiment_count = df["Sentiment"].value_counts()
print(sentiment_count)
```

```python
issue_count = df["Issue"].value_counts()
print(issue_count)
```
